## One-Hot Encoding (in ML / NLP)

One-Hot Encoding is a technique used to convert categorical data into numerical form so that ML models can understand it.

👉 Each category is represented as a binary vector (0s and 1s).

🔹 Example

Category: Color = ["Red", "Blue", "Green"]

One-Hot Encoded:

Red   → [1, 0, 0]
Blue  → [0, 1, 0]
Green → [0, 0, 1]


### Advantage and Disadvantage
#### ✅ Advantages of One-Hot Encoding

    1️⃣ Prevents false ordering of categories
    2️⃣ Easy to understand & implement
    3️⃣ Works well with linear models
    4️⃣ No assumption about category relationship

#### ❌ Disadvantages of One-Hot Encoding

    1️⃣ High dimensionality (curse of dimensionality)
    2️⃣ Uses more memory & computation
    3️⃣ Not suitable for high-cardinality features
    4️⃣ Sparse vectors reduce model efficiency


```python
from sklearn.preprocessing import OneHotEncoder
import numpy as np

data = np.array([['Red'], ['Blue'], ['Green']])

encoder = OneHotEncoder()
encoded = encoder.fit_transform(data)

print(encoded)
```

## Conclusion
🔥 Interview One-Liner

👉 One-hot encoding converts categorical variables into binary vectors without introducing ordinal relationships.

When to use?

✔ Low number of categories
✔ Linear / tree models

When NOT to use?

❌ Too many unique values → use Label Encoding / Embeddings
