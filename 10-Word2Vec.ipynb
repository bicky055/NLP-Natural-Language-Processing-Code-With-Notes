# Word Embedding (in NLP)

Word Embedding is a technique that represents words as dense numerical vectors in such a way that similar words have similar vectors.

👉 Unlike BoW / TF-IDF, embeddings capture meaning (semantics).

## 🔹 Simple Idea

    Words → numbers

    Similar meaning → closer vectors

    Example:
    king – man + woman ≈ queen

## 🔹 Why Word Embeddings?

    1️⃣ Capture context & semantic meaning
    2️⃣ Reduce high dimensionality
    3️⃣ Better performance in Deep Learning models
    4️⃣ Understand relationships between words

## 🔹 Types of Word Embeddings
    1️⃣ Word2Vec

    Predicts words using context

    Two models: CBOW & Skip-Gram

    2️⃣ GloVe

    Uses global word co-occurrence statistics

    3️⃣ FastText

    Uses subwords (handles spelling mistakes)

    4️⃣ Contextual Embeddings

    ELMo, BERT, GPT

    Word meaning depends on context

## ✅ Advantages of Word Embeddings

    1️⃣ Capture semantic similarity
    2️⃣ Dense & memory efficient
    3️⃣ Handle synonyms well
    4️⃣ Better than TF-IDF for DL

## ❌ Disadvantages of Word Embeddings

    1️⃣ Need large data
    2️⃣ Hard to interpret
    3️⃣ Pretrained models may have bias
    4️⃣ Training is computationally expensive

# Word2Vec (in NLP)

Word2Vec is a word embedding technique that converts words into dense vectors by learning from the context of words.

👉 Words with similar meaning get similar vectors.

## 🔹 How Word2Vec works

    It learns word vectors using a neural network with two approaches:

    1️⃣ CBOW (Continuous Bag of Words)

    Predicts the target word using context words

    Faster, works well on large data

    Example:
    I love ___ → predicts NLP

    2️⃣ Skip-Gram

    Predicts context words using the target word

    Better for rare words

    More accurate but slower

    Example:
    NLP → predicts I, love
