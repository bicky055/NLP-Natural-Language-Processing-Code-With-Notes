## Named Entity Recognizer (NER) in NLP

Named Entity Recognition (NER) is an NLP technique used to identify and classify real-world entities in text into predefined categories.

👉 It answers: Who? What? Where? When?

### 🔹 Common Entity Types

    PERSON – names of people

    ORG – organizations

    GPE / LOCATION – places

    DATE / TIME

    MONEY, PERCENT

### Lest Code to understand

```python
sentence="The Eiffel Tower was built from 1887 to 1889 by Gustave Eiffel, whose company specialized in building metal frameworks and structures."
```

```python
import nltk 
words = nltk.word_tokenize(sentence)
```

```python
tag_elements=nltk.pos_tag(words)
```

```python
nltk.download('maxent_ne_chunker')
```
