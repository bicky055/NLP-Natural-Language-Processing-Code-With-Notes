Tokenization in NLP means breaking text into small pieces called tokens.

👉 Tokens can be words, sentences, or characters.

Simple example 👇

Text:
I love Machine Learning

After tokenization:
["I", "love", "Machine", "Learning"]

```python
corpus="""Hello Welcome,to Krish Naik's NLP Tutorials.
Please do watch the entire course! to become expert in NLP.
"""
```

```python
print(corpus)
```

```python
##  Tokenization
## Sentence-->paragraphs
from nltk.tokenize import sent_tokenize
```

```python
documents=sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
##Tokenization
## paragraph --> words
## sentence --> words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer=TreebankWordTokenizer()

tokenizer.tokenize(corpus)
```

```python
# Step 1: Import nltk and tokenizer
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

# Step 2: Download required package (run once)
nltk.download('punkt')
nltk.download('punkt_tab')

# Step 3: Define corpus (text data)
corpus = """Natural Language Processing (NLP) is a part of Artificial Intelligence.
It helps machines understand human language.
NLP is widely used in chatbots, translation, and fake news detection."""

# Step 4: Sentence Tokenization
documents = sent_tokenize(corpus)
print("Sentences:")
print(documents)

# Step 5: Word Tokenization
words = word_tokenize(corpus)
print("\nWords:")
print(words)
```
